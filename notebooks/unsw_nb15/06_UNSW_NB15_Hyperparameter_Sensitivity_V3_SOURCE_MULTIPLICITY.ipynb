```python
# ============================================================
# 4. HASHING, IMMUTABLE JSON, CONTRACT, AND STAGE UTILITIES
# ============================================================
ALLOWED_PASS_STATUSES = {"PASS", "PASS_WITH_NONBLOCKING_WARNINGS"}


def hard_fail(message):
    print("EXPERIMENT 1 v2 MODEL-STAGE INTEGRITY: FAIL")
    print("Next authorized action: NONE until the blocking issue is corrected/versioned.")
    raise RuntimeError(str(message))


def require(condition, message):
    if not bool(condition):
        hard_fail(message)


def canonical_json_bytes(obj):
    return json.dumps(obj, sort_keys=True, separators=(",", ":"), ensure_ascii=False, allow_nan=False).encode("utf-8")


def sha256_bytes(data):
    return hashlib.sha256(data).hexdigest()


def sha256_file(path, block_size=16 * 1024 * 1024):
    h = hashlib.sha256()
    with Path(path).open("rb") as f:
        while True:
            b = f.read(block_size)
            if not b:
                break
            h.update(b)
    return h.hexdigest()


def read_json(path):
    with Path(path).open("r", encoding="utf-8") as f:
        return json.load(f)


def atomic_write_json(path, obj):
    path = Path(path); path.parent.mkdir(parents=True, exist_ok=True)
    tmp = path.with_suffix(path.suffix + ".tmp")
    with tmp.open("w", encoding="utf-8") as f:
        json.dump(obj, f, indent=2, sort_keys=True, ensure_ascii=False, allow_nan=False)
        f.write("\n")
    os.replace(tmp, path)


def atomic_write_csv(path, df):
    path = Path(path); path.parent.mkdir(parents=True, exist_ok=True)
    tmp = path.with_suffix(path.suffix + ".tmp")
    df.to_csv(tmp, index=False)
    os.replace(tmp, path)


def atomic_write_parquet(path, df):
    path = Path(path); path.parent.mkdir(parents=True, exist_ok=True)
    tmp = path.with_suffix(path.suffix + ".tmp")
    df.to_parquet(tmp, index=False, compression="snappy")
    os.replace(tmp, path)


def substantive_hash(obj, excluded_keys):
    body = {k: v for k, v in obj.items() if k not in set(excluded_keys)}
    return sha256_bytes(canonical_json_bytes(body)), body


def verify_self_hashed_json(path, hash_field, extra_excluded=()):
    obj = read_json(path)
    excluded = {hash_field, "created_at_utc", *extra_excluded}
    computed, _ = substantive_hash(obj, excluded)
    stored = str(obj.get(hash_field, "")).lower()
    require(stored and computed.lower() == stored, f"Self-hash mismatch for {path}: computed={computed}, stored={stored}")
    return obj, computed


def freeze_contract(path, substantive, hash_field="contract_hash"):
    path = Path(path)
    expected_hash = sha256_bytes(canonical_json_bytes(substantive))
    if path.exists():
        obj = read_json(path)
        stored = str(obj.get(hash_field, "")).lower()
        body = {k: v for k, v in obj.items() if k not in {hash_field, "created_at_utc"}}
        require(stored == expected_hash.lower(), f"Existing contract hash mismatch: {path}")
        require(body == substantive, f"Existing contract substantive content differs: {path}")
        return obj, expected_hash
    obj = {**substantive, "created_at_utc": datetime.now(timezone.utc).isoformat(), hash_field: expected_hash}
    atomic_write_json(path, obj)
    return obj, expected_hash


def verify_contract(path, hash_field="contract_hash"):
    path = Path(path)
    require(path.exists(), f"Required global contract is missing: {path}. Run Notebook 01 first for this Experiment 1 v2 project.")
    return verify_self_hashed_json(path, hash_field)


def make_stage_manifest(substantive):
    h = sha256_bytes(canonical_json_bytes(substantive))
    return {**substantive, "created_at_utc": datetime.now(timezone.utc).isoformat(), "stage_manifest_hash": h}


def write_stage_manifest(path, substantive):
    obj = make_stage_manifest(substantive)
    path = Path(path)
    if path.exists():
        old, _ = verify_self_hashed_json(path, "stage_manifest_hash")
        old_body = {k: v for k, v in old.items() if k not in {"created_at_utc", "stage_manifest_hash"}}
        require(old_body == substantive, f"Existing stage manifest conflicts with current execution contract: {path}")
        return old
    atomic_write_json(path, obj)
    return obj


def load_stage_manifest(path):
    obj, _ = verify_self_hashed_json(path, "stage_manifest_hash")
    require(obj.get("final_status") in ALLOWED_PASS_STATUSES, f"Upstream stage is not PASS: {path}")
    require(obj.get("dataset") == DATASET_NAME, f"Stage manifest dataset mismatch: {path}")
    return obj


def dataset_matches(value):
    a = str(value).strip().casefold().replace("_", "-")
    b = DATASET_NAME.casefold().replace("_", "-")
    aliases = {b}
    if DATASET_NAME == "CICIDS2017": aliases |= {"cicids-2017", "cicids2017"}
    else: aliases |= {"unsw-nb15", "unswnb15"}
    return a in aliases


def discover_one(paths, description):
    unique = []
    seen = set()
    for p in paths:
        p = Path(p).resolve()
        if str(p) not in seen:
            unique.append(p); seen.add(str(p))
    require(len(unique) == 1, f"Expected exactly one {description}; found {len(unique)}: {[str(x) for x in unique]}")
    return unique[0]


def resolve_dependency_stage(stage_slug, require_formal=False):
    override = STAGE_ROOT_OVERRIDES.get(stage_slug)
    if override:
        root = Path(override).resolve()
        manifest_path = root / "stage_manifest.json"
        manifest = load_stage_manifest(manifest_path)
        for rel, expected in manifest.get("key_artifact_hashes", {}).items():
            ap = root / rel
            require(ap.exists(), f"Frozen upstream stage artifact is missing: {ap}")
            require(sha256_file(ap) == expected, f"Frozen upstream stage artifact hash mismatch: {ap}")
        return root, manifest
    base = MODELLING_ROOT / stage_slug / MODEL_BRANCH_VERSION
    candidates = []
    if base.exists():
        for p in base.rglob("stage_manifest.json"):
            try:
                obj, _ = verify_self_hashed_json(p, "stage_manifest_hash")
                if obj.get("dataset") != DATASET_NAME or obj.get("final_status") not in ALLOWED_PASS_STATUSES:
                    continue
                if require_formal and obj.get("run_mode") != "formal":
                    continue
                candidates.append(p.parent.resolve())
            except Exception:
                continue
    root = discover_one(candidates, f"valid upstream stage root for {stage_slug}")
    manifest = load_stage_manifest(root / "stage_manifest.json")
    _current_upstream_identity = globals().get("UPSTREAM_IDENTITY")
    if _current_upstream_identity is not None and "upstream_identity" in manifest:
        require(
            manifest.get("upstream_identity") == _current_upstream_identity,
            f"Upstream stage {stage_slug} is bound to a different PRE-02/partition identity."
        )

    for rel, expected in manifest.get("key_artifact_hashes", {}).items():
        ap = root / rel
        require(ap.exists(), f"Frozen upstream stage artifact is missing: {ap}")
        require(sha256_file(ap) == expected, f"Frozen upstream stage artifact hash mismatch: {ap}")
    return root, manifest


def create_artifact_inventory(root, output_path):
    root = Path(root); output_path = Path(output_path)
    rows = []
    for p in sorted(root.rglob("*")):
        if not p.is_file() or p.resolve() == output_path.resolve() or p.name.endswith(".tmp"):
            continue
        rows.append({
            "relative_path": str(p.relative_to(root)),
            "size_bytes": int(p.stat().st_size),
            "sha256": sha256_file(p),
        })
    atomic_write_csv(output_path, pd.DataFrame(rows))
    return rows
```

```python
# ============================================================
# 6. FROZEN TRANSFORMED-DATA ACCESS / LOCAL CACHE
# ============================================================
ALLOW_FINAL_TEST_OUTCOMES = False
EXPERIMENTAL_PARTITIONS = [
    "benign_weight_train",
    "benign_early_stop",
    "benign_threshold_calibration",
    "development",
    "final_test",
]
FEATURE_NAMES = list(REPRESENTATION.get("exact_transformed_feature_order", []))
INPUT_DIM = int(REPRESENTATION.get("total_transformed_dimensionality", -1))
TRANSFORMED_DTYPE = str(REPRESENTATION.get("transformed_dtype"))

require(
    INPUT_DIM > 0 and len(FEATURE_NAMES) == INPUT_DIM,
    "Frozen transformed representation has invalid dimension/feature order.",
)
require(
    TRANSFORMED_DTYPE in {"float32", "float64"},
    f"Unsupported frozen transformed dtype: {TRANSFORMED_DTYPE}",
)

NP_DTYPE = np.dtype(TRANSFORMED_DTYPE)
TORCH_DTYPE = torch.float32 if TRANSFORMED_DTYPE == "float32" else torch.float64
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

require(
    set(CORRUPTION.get("categorical_groups", {})) == {"proto", "state", "service"},
    "UNSW corruption groups must be exactly proto/state/service.",
)

LOCAL_CACHE_ROOT = (
    Path("/content/experiment1_v2_model_cache")
    / DATASET_PREFIX
    / REPRESENTATION.get("feature_order_hash", "unknown")
)
LOCAL_CACHE_ROOT.mkdir(parents=True, exist_ok=True)

# Session-level caches eliminate repeated SHA-256 reads of immutable Drive
# artifacts. Every artifact is still cryptographically verified at least once
# in the current runtime before it can be cached/used.
_PARTITION_VERIFICATION_CACHE = {}
_RAM_MATRIX_CACHE = {}


def _partition_paths(partition):
    require(partition in EXPERIMENTAL_PARTITIONS, f"Unknown partition: {partition}")
    if partition == "final_test" and not ALLOW_FINAL_TEST_OUTCOMES:
        hard_fail(
            "Final-test feature/metadata access is forbidden before Notebook 07. "
            f"Attempted partition={partition}"
        )
    return (
        PRE02_ROOT / "transformed_partitions" / f"{partition}_transformed.parquet",
        PRE02_ROOT / "transformed_partitions" / f"{partition}_transformed_alignment.parquet",
        PRE02_ROOT / "raw_partitions" / f"{partition}_raw_metadata.parquet",
    )


def _partition_manifest_info(partition):
    info = REPRESENTATION.get("per_partition", {}).get(partition)
    require(
        isinstance(info, dict),
        f"Representation manifest lacks partition metadata: {partition}",
    )
    return info


def verify_partition_files(partition, verify_model_bytes=True):
    model_path, alignment_path, metadata_path = _partition_paths(partition)
    info = _partition_manifest_info(partition)

    cached = _PARTITION_VERIFICATION_CACHE.get(partition)
    if cached is not None:
        # Model-byte verification is the stronger state. If it already occurred,
        # no Drive re-hashing is needed in this runtime.
        if cached.get("model_verified", False) or not verify_model_bytes:
            for p in [model_path, alignment_path, metadata_path]:
                require(p.exists(), f"Required PRE-02 artifact disappeared: {p}")
            return model_path, alignment_path, metadata_path

    for p in [model_path, alignment_path, metadata_path]:
        require(p.exists(), f"Required PRE-02 artifact is missing: {p}")

    if verify_model_bytes:
        require(
            sha256_file(model_path) == info.get("transformed_artifact_sha256"),
            f"Transformed artifact SHA mismatch for {partition}",
        )

    # Alignment and metadata hashes are checked once per runtime.
    require(
        sha256_file(alignment_path) == info.get("transformed_alignment_sha256"),
        f"Alignment artifact SHA mismatch for {partition}",
    )
    require(
        sha256_file(metadata_path) == info.get("raw_metadata_sha256"),
        f"Raw metadata SHA mismatch for {partition}",
    )

    pf = pq.ParquetFile(model_path)
    require(
        pf.schema_arrow.names == FEATURE_NAMES,
        f"Feature order mismatch for {partition}",
    )
    require(
        int(pf.metadata.num_rows) == int(info.get("transformed_row_count")),
        f"Row count mismatch for {partition}",
    )

    _PARTITION_VERIFICATION_CACHE[partition] = {
        "model_verified": bool(verify_model_bytes),
        "alignment_verified": True,
        "metadata_verified": True,
        "source_hash": info.get("transformed_artifact_sha256"),
    }
    return model_path, alignment_path, metadata_path


def ensure_local_matrix(partition):
    if partition in _RAM_MATRIX_CACHE:
        return _RAM_MATRIX_CACHE[partition]

    model_path, _, _ = verify_partition_files(
        partition,
        verify_model_bytes=True,
    )
    info = _partition_manifest_info(partition)
    rows = int(info["transformed_row_count"])

    cache_path = LOCAL_CACHE_ROOT / f"{partition}.npy"
    meta_path = LOCAL_CACHE_ROOT / f"{partition}.cache.json"
    expected_meta = {
        "source_sha256": info["transformed_artifact_sha256"],
        "rows": rows,
        "columns": INPUT_DIM,
        "dtype": TRANSFORMED_DTYPE,
        "feature_order_hash": REPRESENTATION["feature_order_hash"],
    }

    if cache_path.exists() and meta_path.exists():
        try:
            observed = read_json(meta_path)
            mm = np.load(cache_path, mmap_mode="r")
            if (
                observed == expected_meta
                and mm.shape == (rows, INPUT_DIM)
                and str(mm.dtype) == TRANSFORMED_DTYPE
            ):
                return mm
        except Exception:
            pass
        cache_path.unlink(missing_ok=True)
        meta_path.unlink(missing_ok=True)

    # One-time conversion from Drive Parquet -> local NPY memmap.
    mm = np.lib.format.open_memmap(
        cache_path,
        mode="w+",
        dtype=NP_DTYPE,
        shape=(rows, INPUT_DIM),
    )
    pf = pq.ParquetFile(model_path)
    offset = 0
    for batch in pf.iter_batches(batch_size=50_000, columns=FEATURE_NAMES):
        arr = batch.to_pandas().to_numpy(dtype=NP_DTYPE, copy=False)
        require(
            np.isfinite(arr).all(),
            f"Non-finite value found while caching {partition}",
        )
        mm[offset : offset + len(arr)] = arr
        offset += len(arr)

    require(offset == rows, f"Local cache row count mismatch for {partition}")
    mm.flush()
    del mm
    atomic_write_json(meta_path, expected_meta)
    return np.load(cache_path, mmap_mode="r")


def _available_memory_bytes():
    try:
        with open("/proc/meminfo", "r", encoding="utf-8") as f:
            for line in f:
                if line.startswith("MemAvailable:"):
                    return int(line.split()[1]) * 1024
    except Exception:
        pass
    return 0


def promote_train_early_to_ram_if_safe():
    if not PROMOTE_TRAIN_EARLY_TO_RAM_IF_SAFE:
        return False

    train_mm = ensure_local_matrix("benign_weight_train")
    early_mm = ensure_local_matrix("benign_early_stop")
    needed = int(train_mm.nbytes + early_mm.nbytes)
    available = _available_memory_bytes()

    if available <= 0:
        print("RAM promotion skipped: available memory could not be determined.")
        return False

    limit = int(float(MAX_RAM_FRACTION_FOR_TRAIN_EARLY) * available)
    if needed > limit:
        print(
            "RAM promotion skipped:",
            f"train+early need {needed/1024**3:.2f} GiB; "
            f"allowed operational budget is {limit/1024**3:.2f} GiB.",
        )
        return False

    print(
        "Promoting benign_weight_train + benign_early_stop to RAM:",
        f"{needed/1024**3:.2f} GiB total.",
    )
    _RAM_MATRIX_CACHE["benign_weight_train"] = np.asarray(train_mm).copy()
    _RAM_MATRIX_CACHE["benign_early_stop"] = np.asarray(early_mm).copy()
    return True


def ensure_local_sidecars(partition):
    _, alignment_path, metadata_path = verify_partition_files(
        partition,
        verify_model_bytes=False,
    )
    info = _partition_manifest_info(partition)

    side_root = LOCAL_CACHE_ROOT / "sidecars"
    side_root.mkdir(parents=True, exist_ok=True)

    local_alignment = side_root / f"{partition}_alignment.parquet"
    local_metadata = side_root / f"{partition}_metadata.parquet"
    local_manifest = side_root / f"{partition}_sidecar_cache.json"

    expected = {
        "alignment_sha256": info.get("transformed_alignment_sha256"),
        "metadata_sha256": info.get("raw_metadata_sha256"),
        "row_count": int(info.get("transformed_row_count")),
    }

    valid_local = False
    if local_alignment.exists() and local_metadata.exists() and local_manifest.exists():
        try:
            valid_local = read_json(local_manifest) == expected
        except Exception:
            valid_local = False

    if not valid_local:
        local_alignment.unlink(missing_ok=True)
        local_metadata.unlink(missing_ok=True)
        local_manifest.unlink(missing_ok=True)

        tmp_a = local_alignment.with_suffix(".parquet.tmp")
        tmp_m = local_metadata.with_suffix(".parquet.tmp")
        shutil.copy2(alignment_path, tmp_a)
        shutil.copy2(metadata_path, tmp_m)
        os.replace(tmp_a, local_alignment)
        os.replace(tmp_m, local_metadata)

        # Validate local copies before declaring the cache usable.
        require(
            sha256_file(local_alignment) == expected["alignment_sha256"],
            f"Local alignment cache hash mismatch for {partition}",
        )
        require(
            sha256_file(local_metadata) == expected["metadata_sha256"],
            f"Local metadata cache hash mismatch for {partition}",
        )

        alignment = pd.read_parquet(
            local_alignment,
            columns=["row_id", "partition"],
        )
        metadata = pd.read_parquet(
            local_metadata,
            columns=["row_id", "partition"],
        )
        require(
            len(alignment) == len(metadata) == expected["row_count"],
            f"Local sidecar row-count mismatch for {partition}",
        )
        require(
            alignment["row_id"].astype(str).reset_index(drop=True).equals(
                metadata["row_id"].astype(str).reset_index(drop=True)
            ),
            f"Local sidecar row_id mismatch for {partition}",
        )
        del alignment, metadata
        gc.collect()

        atomic_write_json(local_manifest, expected)

    return local_alignment, local_metadata


def load_partition_metadata(partition):
    alignment_path, metadata_path = ensure_local_sidecars(partition)

    # Alignment identity was already verified when the local sidecar cache was
    # created. Read only metadata for repeated run-level scoring.
    metadata = pd.read_parquet(metadata_path)
    require(
        (metadata["partition"].astype(str) == partition).all(),
        f"Metadata partition label mismatch for {partition}",
    )
    require(
        "binary_label" in metadata.columns,
        f"binary_label missing from metadata for {partition}",
    )

    y = pd.to_numeric(metadata["binary_label"], errors="raise").astype(int)
    require(
        set(y.unique()).issubset({0, 1}),
        f"Unexpected binary labels in {partition}: {sorted(y.unique())}",
    )

    metadata = metadata.copy()
    metadata["binary_label"] = y
    return metadata.reset_index(drop=True)


def verify_role_contracts(include_final=False):
    rows = []
    partitions = [
        "benign_weight_train",
        "benign_early_stop",
        "benign_threshold_calibration",
        "development",
    ] + (["final_test"] if include_final else [])

    for p in partitions:
        meta = load_partition_metadata(p)
        attacks = int(meta["binary_label"].sum())
        if p in {
            "benign_weight_train",
            "benign_early_stop",
            "benign_threshold_calibration",
        }:
            require(attacks == 0, f"{p} is not benign-only.")
        if p in {"development", "final_test"}:
            require(
                meta["binary_label"].nunique() == 2,
                f"{p} must contain benign and attack observations.",
            )
        rows.append(
            {
                "partition": p,
                "rows": len(meta),
                "attack_rows": attacks,
                "benign_rows": len(meta) - attacks,
            }
        )
        del meta
        gc.collect()

    display(pd.DataFrame(rows))
    return rows


def prewarm_nonfinal_cache():
    """
    Cache only development-authorized partitions. final_test remains inaccessible.
    """
    if not PREWARM_NONFINAL_LOCAL_CACHE:
        return

    print("Prewarming local PRE-02 caches for Notebook 06...")
    for p in [
        "benign_weight_train",
        "benign_early_stop",
        "benign_threshold_calibration",
        "development",
    ]:
        mm = ensure_local_matrix(p)
        ensure_local_sidecars(p)
        print(
            f"  cached {p}: rows={len(mm):,}, features={INPUT_DIM:,}, "
            f"dtype={mm.dtype}"
        )
        del mm
        gc.collect()

    promote_train_early_to_ram_if_safe()


print("Frozen input dimension:", INPUT_DIM)
print("Frozen transformed dtype:", TRANSFORMED_DTYPE)
print("Model device:", DEVICE)
if DEVICE.type != "cuda":
    print(
        "[PERFORMANCE WARNING] CUDA GPU is not active. Notebook 06 can take "
        "many tens of hours on CPU. A Colab GPU runtime is strongly recommended."
    )
```

```python
# ============================================================
# 7. MODEL, TRAINING, CORRUPTION, THRESHOLD, AND METRIC UTILITIES
# ============================================================
ARCHITECTURES = MODEL_SEARCH_CONTRACT["architectures"]
FIXED_TRAINING = MODEL_SEARCH_CONTRACT["fixed_training"]
_adam_params = set(inspect.signature(torch.optim.AdamW).parameters)
for _k in FIXED_TRAINING.get("adamw_fixed_defaults", {}):
    require(_k in _adam_params, f"Frozen AdamW argument {_k!r} is unsupported by this PyTorch runtime; do not silently substitute defaults.")
_sched_params = set(inspect.signature(torch.optim.lr_scheduler.ReduceLROnPlateau).parameters)
for _k in FIXED_TRAINING.get("reduce_lr_on_plateau_fixed_defaults", {}):
    require(_k in _sched_params, f"Frozen scheduler argument {_k!r} is unsupported by this PyTorch runtime; do not silently substitute defaults.")


def set_global_seed(seed):
    seed = int(seed)
    os.environ["PYTHONHASHSEED"] = str(seed)
    random.seed(seed); np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)
    # Deterministic algorithms are requested where available; some CUDA operations
    # may still depend on runtime/library support and this limitation is recorded.
    try:
        torch.use_deterministic_algorithms(True, warn_only=True)
    except Exception:
        pass
    if hasattr(torch.backends, "cudnn"):
        torch.backends.cudnn.deterministic = True
        torch.backends.cudnn.benchmark = False


def latent_dim_for(architecture, latent_fraction):
    hidden = ARCHITECTURES[str(architecture)]
    final_width = int(hidden[-1])
    z = max(1, int(round(final_width * float(latent_fraction))))
    require(z < INPUT_DIM, f"Latent dimension must be undercomplete: z={z}, p={INPUT_DIM}")
    return z


class Autoencoder(nn.Module):
    def __init__(self, input_dim, hidden_widths, latent_dim):
        super().__init__()
        enc = []
        prev = int(input_dim)
        for width in hidden_widths:
            layer = nn.Linear(prev, int(width)); nn.init.kaiming_normal_(layer.weight, mode="fan_in", nonlinearity="relu"); nn.init.zeros_(layer.bias)
            enc += [layer, nn.ReLU()]; prev = int(width)
        latent = nn.Linear(prev, int(latent_dim)); nn.init.kaiming_normal_(latent.weight, mode="fan_in", nonlinearity="relu"); nn.init.zeros_(latent.bias)
        enc += [latent]
        self.encoder = nn.Sequential(*enc)

        dec = []
        prev = int(latent_dim)
        for width in reversed(hidden_widths):
            layer = nn.Linear(prev, int(width)); nn.init.kaiming_normal_(layer.weight, mode="fan_in", nonlinearity="relu"); nn.init.zeros_(layer.bias)
            dec += [layer, nn.ReLU()]; prev = int(width)
        out = nn.Linear(prev, int(input_dim)); nn.init.kaiming_normal_(out.weight, mode="fan_in", nonlinearity="relu"); nn.init.zeros_(out.bias)
        dec += [out]
        self.decoder = nn.Sequential(*dec)

    def forward(self, x):
        return self.decoder(self.encoder(x))


def build_model(config):
    return Autoencoder(INPUT_DIM, ARCHITECTURES[config["architecture"]], int(config["latent_dim"])).to(device=DEVICE, dtype=TORCH_DTYPE)


def parameter_count(model):
    return int(sum(p.numel() for p in model.parameters() if p.requires_grad))


def apply_structured_corruption(clean_x, config):
    x = clean_x.clone()
    numerical_p = float(config.get("numerical_masking", 0.0) or 0.0)
    numerical_indices = list(map(int, CORRUPTION.get("numerical_indices", [])))
    if numerical_p > 0 and numerical_indices:
        idx = torch.tensor(numerical_indices, dtype=torch.long, device=x.device)
        mask = torch.rand((x.shape[0], len(numerical_indices)), device=x.device) < numerical_p
        vals = x.index_select(1, idx)
        vals = torch.where(mask, torch.zeros_like(vals), vals)
        x[:, idx] = vals
    cat_p = float(config.get("grouped_categorical_masking", 0.0) or 0.0)
    if cat_p > 0:
        groups = CORRUPTION.get("categorical_groups", {})
        require(bool(groups), "Categorical masking requested but corruption manifest contains no categorical groups.")
        for _, group in groups.items():
            idx = torch.tensor(list(map(int, group)), dtype=torch.long, device=x.device)
            group_mask = (torch.rand((x.shape[0], 1), device=x.device) < cat_p)
            vals = x.index_select(1, idx)
            vals = torch.where(group_mask, torch.zeros_like(vals), vals)
            x[:, idx] = vals
    return x


def training_loss_fn(name):
    if name == "MSE": return nn.MSELoss(reduction="mean")
    if name == "Huber": return nn.HuberLoss(delta=float(FIXED_TRAINING["huber_delta"]), reduction="mean")
    hard_fail(f"Unsupported training loss: {name}")


def iter_sequential_batches(mm, batch_size):
    for start in range(0, len(mm), int(batch_size)):
        arr = np.asarray(mm[start:start+int(batch_size)])
        yield torch.tensor(arr, dtype=TORCH_DTYPE, device=DEVICE)


def clean_validation_loss(model, mm, batch_size, criterion):
    model.eval(); total = 0.0; n = 0
    with torch.no_grad():
        for xb in iter_sequential_batches(mm, batch_size):
            pred = model(xb); loss = criterion(pred, xb)
            total += float(loss.item()) * len(xb); n += len(xb)
    return total / max(n, 1)


def _atomic_torch_save(path, obj):
    path = Path(path)
    path.parent.mkdir(parents=True, exist_ok=True)
    tmp = path.with_suffix(path.suffix + ".tmp")
    torch.save(obj, tmp)
    os.replace(tmp, path)


def _capture_rng_state(rng):
    state = {
        "python_random_state": random.getstate(),
        "numpy_global_state": np.random.get_state(),
        "numpy_generator_state": rng.bit_generator.state,
        "torch_cpu_rng_state": torch.get_rng_state(),
    }
    if torch.cuda.is_available():
        state["torch_cuda_rng_state_all"] = torch.cuda.get_rng_state_all()
    return state


def _restore_rng_state(state, rng):
    random.setstate(state["python_random_state"])
    np.random.set_state(state["numpy_global_state"])
    rng.bit_generator.state = state["numpy_generator_state"]
    torch.set_rng_state(state["torch_cpu_rng_state"])
    if torch.cuda.is_available() and "torch_cuda_rng_state_all" in state:
        torch.cuda.set_rng_state_all(state["torch_cuda_rng_state_all"])


def train_model(
    config,
    model_type,
    seed,
    checkpoint_path,
    resume_state_path=None,
):
    set_global_seed(seed)
    train_mm = ensure_local_matrix("benign_weight_train")
    early_mm = ensure_local_matrix("benign_early_stop")

    model = build_model(config)
    pcount = parameter_count(model)

    _adam_kwargs = dict(FIXED_TRAINING.get("adamw_fixed_defaults", {}))
    if "betas" in _adam_kwargs:
        _adam_kwargs["betas"] = tuple(_adam_kwargs["betas"])

    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=float(config["learning_rate"]),
        weight_decay=float(config["weight_decay"]),
        **_adam_kwargs,
    )

    _sched_kwargs = dict(
        FIXED_TRAINING.get("reduce_lr_on_plateau_fixed_defaults", {})
    )
    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
        optimizer,
        mode="min",
        factor=float(FIXED_TRAINING["lr_reduction_factor"]),
        patience=int(FIXED_TRAINING["lr_scheduler_patience"]),
        min_lr=float(FIXED_TRAINING["minimum_learning_rate"]),
        **_sched_kwargs,
    )

    criterion = training_loss_fn(config["training_loss"])
    rng = np.random.default_rng(int(seed))

    best_val = float("inf")
    best_epoch = -1
    stale = 0
    history = []
    start_epoch = 1
    prior_elapsed = 0.0

    checkpoint_path = Path(checkpoint_path)
    checkpoint_path.parent.mkdir(parents=True, exist_ok=True)

    resume_state_path = (
        Path(resume_state_path)
        if resume_state_path is not None
        else None
    )

    # Exact epoch-boundary continuation after a future runtime disconnect.
    if (
        resume_state_path is not None
        and resume_state_path.exists()
        and SAVE_EPOCH_RESUME_STATE
    ):
        state = torch.load(resume_state_path, map_location=DEVICE)
        require(
            state.get("config") == config,
            "Resume-state configuration does not match the requested run.",
        )
        require(
            str(state.get("model_type")) == str(model_type),
            "Resume-state model type mismatch.",
        )
        require(
            int(state.get("seed")) == int(seed),
            "Resume-state seed mismatch.",
        )

        model.load_state_dict(state["model_state_dict"])
        optimizer.load_state_dict(state["optimizer_state_dict"])
        scheduler.load_state_dict(state["scheduler_state_dict"])

        best_val = float(state["best_val"])
        best_epoch = int(state["best_epoch"])
        stale = int(state["stale"])
        history = list(state["history"])
        prior_elapsed = float(state.get("elapsed_training_seconds", 0.0))
        start_epoch = int(state["last_completed_epoch"]) + 1

        _restore_rng_state(state["rng_state"], rng)

        print(
            "Resuming interrupted training at epoch",
            start_epoch,
            "from",
            resume_state_path,
        )

    started = time.perf_counter()
    max_epochs = int(FIXED_TRAINING["maximum_epochs"])
    patience = int(FIXED_TRAINING["early_stopping_patience"])

    for epoch in range(start_epoch, max_epochs + 1):
        model.train()
        order = rng.permutation(len(train_mm))
        total = 0.0
        n = 0
        bs = int(config["batch_size"])

        for batch_start in range(0, len(order), bs):
            idx = order[batch_start : batch_start + bs]
            clean = torch.tensor(
                np.asarray(train_mm[idx]),
                dtype=TORCH_DTYPE,
                device=DEVICE,
            )
            inp = (
                apply_structured_corruption(clean, config)
                if model_type == "DAE"
                else clean
            )

            optimizer.zero_grad(set_to_none=True)
            pred = model(inp)
            loss = criterion(pred, clean)
            loss.backward()
            torch.nn.utils.clip_grad_norm_(
                model.parameters(),
                max_norm=float(FIXED_TRAINING["gradient_clip_norm"]),
            )
            optimizer.step()

            total += float(loss.item()) * len(clean)
            n += len(clean)

        train_loss = total / max(n, 1)
        val_loss = clean_validation_loss(
            model,
            early_mm,
            bs,
            criterion,
        )
        scheduler.step(val_loss)
        lr_now = float(optimizer.param_groups[0]["lr"])

        history.append(
            {
                "epoch": epoch,
                "training_loss": train_loss,
                "early_stop_validation_loss": val_loss,
                "learning_rate": lr_now,
            }
        )

        improved = val_loss < best_val
        if improved:
            best_val = val_loss
            best_epoch = epoch
            stale = 0
            _atomic_torch_save(
                checkpoint_path,
                {
                    "state_dict": model.state_dict(),
                    "config": config,
                    "model_type": model_type,
                    "seed": int(seed),
                    "best_epoch": int(best_epoch),
                    "best_validation_loss": float(best_val),
                    "input_dim": INPUT_DIM,
                    "feature_order_hash": REPRESENTATION[
                        "feature_order_hash"
                    ],
                },
            )
        else:
            stale += 1

        elapsed_total = (
            prior_elapsed
            + float(time.perf_counter() - started)
        )

        if resume_state_path is not None and SAVE_EPOCH_RESUME_STATE:
            _atomic_torch_save(
                resume_state_path,
                {
                    "config": config,
                    "model_type": model_type,
                    "seed": int(seed),
                    "last_completed_epoch": int(epoch),
                    "model_state_dict": model.state_dict(),
                    "optimizer_state_dict": optimizer.state_dict(),
                    "scheduler_state_dict": scheduler.state_dict(),
                    "best_val": float(best_val),
                    "best_epoch": int(best_epoch),
                    "stale": int(stale),
                    "history": history,
                    "elapsed_training_seconds": elapsed_total,
                    "rng_state": _capture_rng_state(rng),
                    "input_dim": INPUT_DIM,
                    "feature_order_hash": REPRESENTATION[
                        "feature_order_hash"
                    ],
                },
            )

        if stale >= patience:
            break

    duration = (
        prior_elapsed
        + float(time.perf_counter() - started)
    )

    require(
        checkpoint_path.exists(),
        "Training completed without a best checkpoint.",
    )
    ckpt = torch.load(checkpoint_path, map_location=DEVICE)
    model.load_state_dict(ckpt["state_dict"])

    return (
        model,
        pd.DataFrame(history),
        int(best_epoch),
        float(best_val),
        duration,
        pcount,
    )
def reconstruction_errors(model, mm, batch_size):
    model.eval(); parts = []
    with torch.no_grad():
        for xb in iter_sequential_batches(mm, batch_size):
            pred = model(xb)
            err = torch.mean((xb - pred) ** 2, dim=1)
            parts.append(err.detach().cpu().numpy().astype(np.float64, copy=False))
    return np.concatenate(parts) if parts else np.empty(0, dtype=np.float64)


def quantile_threshold(scores, percentile):
    return float(np.quantile(np.asarray(scores, dtype=np.float64), float(percentile)/100.0, method=THRESHOLD_CONTRACT["quantile_method"]))


def predict_from_threshold(scores, threshold):
    require(THRESHOLD_CONTRACT["decision_operator"] == ">", "This implementation expects the frozen '>' decision operator.")
    return (np.asarray(scores) > float(threshold)).astype(np.int8)


def compute_metrics(y_true, scores, threshold):
    y = np.asarray(y_true, dtype=int); scores = np.asarray(scores, dtype=np.float64)
    pred = predict_from_threshold(scores, threshold)
    require(set(np.unique(y)).issubset({0,1}), "Metrics received non-binary labels.")
    tp = int(np.sum((y==1)&(pred==1))); tn = int(np.sum((y==0)&(pred==0)))
    fp = int(np.sum((y==0)&(pred==1))); fn = int(np.sum((y==1)&(pred==0)))
    n_benign = int(np.sum(y==0))
    fpr = fp / n_benign if n_benign else float("nan")
    precision_curve, recall_curve, _ = precision_recall_curve(y, scores)
    pr_auc = float(auc(recall_curve, precision_curve))
    roc_auc = float(roc_auc_score(y, scores)) if len(np.unique(y)) == 2 else float("nan")
    ap = float(average_precision_score(y, scores)) if len(np.unique(y)) == 2 else float("nan")
    if len(np.unique(y)) == 2:
        fpr_curve, tpr_curve, _ = roc_curve(y, scores)
        eligible = tpr_curve[fpr_curve <= 0.01]
        tpr1 = float(np.max(eligible)) if len(eligible) else 0.0
    else:
        tpr1 = float("nan")
    return {
        "MCC": float(matthews_corrcoef(y, pred)),
        "Recall": float(recall_score(y, pred, zero_division=0)),
        "F1": float(f1_score(y, pred, zero_division=0)),
        "Precision": float(precision_score(y, pred, zero_division=0)),
        "PR-AUC": pr_auc,
        "FPR": float(fpr),
        "FP_per_1000_benign": float(1000.0 * fp / n_benign) if n_benign else float("nan"),
        "Balanced_Accuracy": float(balanced_accuracy_score(y, pred)),
        "ROC-AUC": roc_auc,
        "Average_Precision": ap,
        "TPR_at_1pct_FPR": tpr1,
        "TP": tp, "TN": tn, "FP": fp, "FN": fn,
        "threshold": float(threshold),
    }


def family_diagnostics(metadata, predictions):
    df = pd.DataFrame({
        "y_true": metadata["binary_label"].astype(int).to_numpy(),
        "y_pred": np.asarray(predictions, dtype=int),
        "attack_family": metadata.get("attack_family", pd.Series([None]*len(metadata))).astype("string").to_numpy(),
    })
    attacks = df[df["y_true"] == 1].copy()
    require(len(attacks) > 0, "No attacks are available for family diagnostics.")
    require(not pd.isna(attacks["attack_family"]).any(), "Attack-family metadata is missing for one or more attack observations.")
    total_attack = len(attacks); total_tp = int((attacks["y_pred"]==1).sum()); total_fn = int((attacks["y_pred"]==0).sum())
    rows = []
    for family, g in attacks.groupby("attack_family", sort=True, dropna=False):
        support = len(g); tp = int((g["y_pred"]==1).sum()); fn = int((g["y_pred"]==0).sum())
        rows.append({
            "attack_family": str(family), "support": support, "TP": tp, "FN": fn,
            "recall": float(tp/support) if support else float("nan"),
            "attack_share": float(support/total_attack),
            "detected_share": float(tp/total_tp) if total_tp else float("nan"),
            "missed_share": float(fn/total_fn) if total_fn else float("nan"),
        })
    return pd.DataFrame(rows)

print("PASS: model/training/threshold/metric utilities loaded. Anomaly score is fixed per-observation reconstruction MSE.")
```

```python
# ============================================================
# 8. CONFIG HASH, RUN HASH, RUN PERSISTENCE, EXECUTION
# ============================================================
def normalized_effective_config(model_type, params):
    c = {
        "model_type": str(model_type),
        "architecture": str(params["architecture"]),
        "latent_fraction": float(params["latent_fraction"]),
        "latent_dim": int(params.get("latent_dim", latent_dim_for(params["architecture"], params["latent_fraction"]))),
        "learning_rate": float(params["learning_rate"]),
        "batch_size": int(params["batch_size"]),
        "weight_decay": float(params["weight_decay"]),
        "training_loss": str(params["training_loss"]),
        "optimizer": "AdamW",
        "hidden_activation": "ReLU",
        "output_activation": "Linear",
        "weight_initialization": "Kaiming_normal",
        "maximum_epochs": int(FIXED_TRAINING["maximum_epochs"]),
        "early_stopping_patience": int(FIXED_TRAINING["early_stopping_patience"]),
        "lr_scheduler_patience": int(FIXED_TRAINING["lr_scheduler_patience"]),
        "lr_reduction_factor": float(FIXED_TRAINING["lr_reduction_factor"]),
        "minimum_learning_rate": float(FIXED_TRAINING["minimum_learning_rate"]),
        "gradient_clip_norm": float(FIXED_TRAINING["gradient_clip_norm"]),
        "huber_delta": float(FIXED_TRAINING["huber_delta"]),
        "adamw_fixed_defaults": FIXED_TRAINING.get("adamw_fixed_defaults", {}),
        "reduce_lr_on_plateau_fixed_defaults": FIXED_TRAINING.get("reduce_lr_on_plateau_fixed_defaults", {}),
        "kaiming_normal_details": FIXED_TRAINING.get("kaiming_normal_details", {}),
        "numerical_masking": float(params.get("numerical_masking", 0.0) or 0.0),
        "corruption_group_manifest_hash": CORRUPTION_HASH,
    }
    if DATASET_NAME == "UNSW-NB15" and model_type in {"DAE", "AE_MATCHED"}:
        c["grouped_categorical_masking"] = float(params.get("grouped_categorical_masking", 0.0) or 0.0)
    return c


def config_hash(config):
    return sha256_bytes(canonical_json_bytes(config))


def make_run_hash(stage, model_type, config_h, seed, attempt=1):
    payload = {
        "dataset": DATASET_NAME, "stage": stage, "model_type": model_type,
        "config_hash": config_h, "seed": int(seed),
        "partition_manifest_hash": UPSTREAM_HASHES.get("partition_manifest_canonical_hash"),
        "preprocessing_manifest_hash": PREPROCESSING_POLICY_HASH,
        "transformed_representation_hash": REPRESENTATION_HASH,
        "corruption_group_manifest_hash": CORRUPTION_HASH,
        "code_version": CODE_VERSION, "notebook_version": NOTEBOOK_VERSION,
        "attempt": int(attempt),
    }
    return sha256_bytes(canonical_json_bytes(payload))


def environment_for_run():
    return {**ENVIRONMENT_INFO, "device": str(DEVICE), "torch_dtype": str(TORCH_DTYPE), "code_version": CODE_VERSION, "notebook_version": NOTEBOOK_VERSION}


def _find_existing_complete_run(stage_root, stage, model_type, cfg_hash, seed):
    runs_dir = Path(stage_root) / "runs"
    if not runs_dir.exists(): return None
    for p in runs_dir.glob("*/run_manifest.json"):
        try:
            obj, _ = verify_self_hashed_json(p, "run_manifest_hash")
            if (obj.get("completion_status") == "COMPLETE" and obj.get("stage") == stage and obj.get("model_type") == model_type
                and obj.get("config_hash") == cfg_hash and int(obj.get("seed")) == int(seed)
                and obj.get("upstream_identity") == UPSTREAM_IDENTITY):
                return p.parent, obj
        except Exception:
            continue
    return None


def execute_training_run(stage_root, stage, model_type, params, seed, evaluation_partition="development", allow_reuse=True):
    require(evaluation_partition in {"development", "final_test"}, "Training-run evaluation partition must be development or final_test.")
    if evaluation_partition == "final_test": require(ALLOW_FINAL_TEST_OUTCOMES, "Final-test outcomes are sealed before Notebook 07.")
    config = normalized_effective_config(model_type, params)
    cfg_hash = config_hash(config)
    if allow_reuse:
        found = _find_existing_complete_run(stage_root, stage, model_type, cfg_hash, seed)
        if found is not None:
            run_dir, manifest = found
            print("Reusing validated complete run:", run_dir.name)
            return manifest

    runs_dir = Path(stage_root) / "runs"; runs_dir.mkdir(parents=True, exist_ok=True)
    attempts = []
    for pattern in ["*/run_manifest.json", "*/failure_manifest.json"]:
        for p in runs_dir.glob(pattern):
            try:
                obj = read_json(p)
                if obj.get("stage") == stage and obj.get("model_type") == model_type and obj.get("config_hash") == cfg_hash and int(obj.get("seed", -1)) == int(seed):
                    attempts.append(int(obj.get("attempt", 1)))
            except Exception:
                pass
    attempt = max(attempts, default=0) + 1
    resume_existing_attempt = False

    while True:
        r_hash = make_run_hash(stage, model_type, cfg_hash, seed, attempt)
        run_dir = runs_dir / r_hash

        if not run_dir.exists():
            break

        resume_state_path = run_dir / "resume_state.pt"
        effective_path = run_dir / "effective_config.json"
        environment_path = run_dir / "environment.json"

        can_resume = (
            SAVE_EPOCH_RESUME_STATE
            and resume_state_path.exists()
            and effective_path.exists()
            and environment_path.exists()
        )

        if can_resume:
            try:
                old_cfg = read_json(effective_path)
                old_env = read_json(environment_path)
                current_env = environment_for_run()

                same_cfg = (
                    old_cfg.get("config_hash") == cfg_hash
                    and old_cfg.get("effective_config") == config
                )
                # Resume only across materially compatible runtime stacks.
                same_runtime = (
                    old_env.get("torch") == current_env.get("torch")
                    and old_env.get("cuda_version") == current_env.get("cuda_version")
                    and old_env.get("device") == current_env.get("device")
                )

                if same_cfg and same_runtime:
                    resume_existing_attempt = True
                    print(
                        "Found resumable interrupted run; continuing the same "
                        f"attempt from its last completed epoch: {run_dir.name}"
                    )
                    break
            except Exception:
                pass

        print(
            "Detected non-resumable incomplete run directory; preserving it "
            f"and advancing attempt: {run_dir.name}"
        )
        attempt += 1

    if not resume_existing_attempt:
        run_dir.mkdir(parents=True)
        atomic_write_json(
            run_dir / "effective_config.json",
            {
                "config_hash": cfg_hash,
                "effective_config": config,
            },
        )
        atomic_write_json(
            run_dir / "environment.json",
            environment_for_run(),
        )

    checkpoint_path = run_dir / "checkpoint.pt"
    resume_state_path = run_dir / "resume_state.pt"

    try:
        model, history, best_epoch, best_val, duration, pcount = train_model(
            config,
            "DAE" if model_type == "DAE" else "AE",
            seed,
            checkpoint_path,
            resume_state_path=resume_state_path,
        )
        atomic_write_csv(run_dir / "training_history.csv", history)
        calibration_mm = ensure_local_matrix("benign_threshold_calibration")
        calibration_scores = reconstruction_errors(model, calibration_mm, config["batch_size"])
        calibration_meta = load_partition_metadata("benign_threshold_calibration")
        require(len(calibration_scores) == len(calibration_meta), "Calibration score/metadata length mismatch.")
        cal_df = pd.DataFrame({"row_id": calibration_meta["row_id"].astype(str), "reconstruction_error": calibration_scores})
        atomic_write_parquet(run_dir / "calibration_scores.parquet", cal_df)
        threshold = quantile_threshold(calibration_scores, THRESHOLD_CONTRACT["primary_percentile"])
        threshold_substantive = {
            "percentile": float(THRESHOLD_CONTRACT["primary_percentile"]), "threshold": threshold,
            "calibration_partition": "benign_threshold_calibration",
            "calibration_partition_hash": REPRESENTATION["per_partition"]["benign_threshold_calibration"]["transformed_artifact_sha256"],
            "calibration_observations": int(len(calibration_scores)),
            "quantile_implementation": THRESHOLD_CONTRACT["quantile_implementation"], "quantile_method": THRESHOLD_CONTRACT["quantile_method"],
            "score_definition": MODEL_SEARCH_CONTRACT["reconstruction_score"], "decision_operator": THRESHOLD_CONTRACT["decision_operator"],
        }
        threshold_manifest = {**threshold_substantive, "threshold_manifest_hash": sha256_bytes(canonical_json_bytes(threshold_substantive))}
        atomic_write_json(run_dir / "threshold_manifest.json", threshold_manifest)

        eval_mm = ensure_local_matrix(evaluation_partition)
        eval_scores = reconstruction_errors(model, eval_mm, config["batch_size"])
        eval_meta = load_partition_metadata(evaluation_partition)
        require(len(eval_scores) == len(eval_meta), "Evaluation score/metadata length mismatch.")
        y_true = eval_meta["binary_label"].astype(int).to_numpy()
        metrics = compute_metrics(y_true, eval_scores, threshold)
        preds = predict_from_threshold(eval_scores, threshold)
        out = pd.DataFrame({
            "row_id": eval_meta["row_id"].astype(str), "y_true": y_true,
            "reconstruction_error": eval_scores, "threshold": float(threshold), "y_pred": preds,
            "model": model_type, "seed": int(seed), "config_hash": cfg_hash, "run_hash": r_hash,
            "attack_family": eval_meta.get("attack_family", pd.Series([None]*len(eval_meta))).astype("string"),
        })
        if "feature_fingerprint" in eval_meta.columns:
            out["feature_fingerprint"] = eval_meta["feature_fingerprint"].astype("string")
        atomic_write_parquet(run_dir / "evaluation_predictions.parquet", out)
        atomic_write_json(run_dir / "metrics.json", metrics)
        checkpoint_hash = sha256_file(checkpoint_path)
        run_substantive = {
            "dataset": DATASET_NAME, "stage": stage, "model_type": model_type, "seed": int(seed),
            "attempt": int(attempt), "effective_config": config, "config_hash": cfg_hash, "run_hash": r_hash,
            "upstream_identity": UPSTREAM_IDENTITY, "evaluation_partition": evaluation_partition,
            "checkpoint_path": str(checkpoint_path), "checkpoint_sha256": checkpoint_hash,
            "best_epoch": int(best_epoch), "best_early_stop_loss": float(best_val),
            "training_duration_seconds": duration, "parameter_count": pcount,
            "threshold": float(threshold), "metrics": metrics,
            "artifact_hashes": {
                "training_history": sha256_file(run_dir / "training_history.csv"),
                "calibration_scores": sha256_file(run_dir / "calibration_scores.parquet"),
                "evaluation_predictions": sha256_file(run_dir / "evaluation_predictions.parquet"),
                "threshold_manifest": sha256_file(run_dir / "threshold_manifest.json"),
                "metrics": sha256_file(run_dir / "metrics.json"),
                "environment": sha256_file(run_dir / "environment.json"),
            },
            "completion_status": "COMPLETE",
        }
        manifest = {**run_substantive, "created_at_utc": datetime.now(timezone.utc).isoformat(), "run_manifest_hash": sha256_bytes(canonical_json_bytes(run_substantive))}
        atomic_write_json(run_dir / "run_manifest.json", manifest)
        # A COMPLETE run no longer needs its in-progress epoch resume state.
        resume_state_path.unlink(missing_ok=True)
        return manifest
    except Exception as exc:
        failure = {
            "dataset": DATASET_NAME, "stage": stage, "model_type": model_type, "seed": int(seed), "attempt": int(attempt),
            "config_hash": cfg_hash, "run_hash": r_hash, "effective_config": config,
            "failure_type": type(exc).__name__, "failure_message": str(exc), "traceback": traceback.format_exc(),
            "created_at_utc": datetime.now(timezone.utc).isoformat(), "completion_status": "FAILED",
        }
        atomic_write_json(run_dir / "failure_manifest.json", failure)
        raise


def collect_complete_run_manifests(stage_root, stage=None, model_type=None):
    rows = []
    for p in (Path(stage_root) / "runs").glob("*/run_manifest.json") if (Path(stage_root)/"runs").exists() else []:
        try:
            obj, _ = verify_self_hashed_json(p, "run_manifest_hash")
            if obj.get("completion_status") != "COMPLETE" or obj.get("upstream_identity") != UPSTREAM_IDENTITY: continue
            if stage is not None and obj.get("stage") != stage: continue
            if model_type is not None and obj.get("model_type") != model_type: continue
            rows.append(obj)
        except Exception:
            continue
    return rows


def run_manifest_to_row(m):
    row = {
        "model_type": m["model_type"], "seed": int(m["seed"]), "config_hash": m["config_hash"], "run_hash": m["run_hash"],
        "parameter_count": int(m["parameter_count"]), "training_duration_seconds": float(m["training_duration_seconds"]),
        "best_epoch": int(m["best_epoch"]), "threshold": float(m["threshold"]),
    }
    row.update(m["metrics"])
    return row
```

## Consolidated UNSW AE + DAE sensitivity execution

This Notebook 06 executes both selected-model sensitivity analyses in one
auditable stage and writes one consolidated AE+DAE result set.

**Scientific design is unchanged.** Consolidation does not reduce the required
number of model trainings; it removes avoidable repeated I/O and makes recovery
from Colab interruptions safer.

Performance/recovery improvements:

- PRE-02 transformed artifacts are SHA-256 verified once per runtime rather than
  repeatedly for every sensitivity run.
- Non-final transformed matrices and metadata are cached on local Colab storage.
- `benign_weight_train` and `benign_early_stop` are promoted to RAM only when a
  bounded memory check says this is safe.
- Completed sensitivity runs are reused.
- Future interrupted runs persist full epoch-boundary optimizer/scheduler/RNG
  state and can resume from the last completed epoch when the runtime stack is
  compatible.
- `final_test` remains sealed and is never cached or accessed here.

The selected AE/DAE configurations remain frozen; sensitivity results do not
reopen model selection.


```python
# ============================================================
# 10. NOTEBOOK 06 — POST-SELECTION ONE-FACTOR HYPERPARAMETER SENSITIVITY
# ============================================================
CONF_ROOT, CONF_STAGE = resolve_dependency_stage("notebook03_multiseed_confirmation")
SELECTED_PATH = CONF_ROOT / "selected_models_manifest.json"
require(SELECTED_PATH.exists(), "Selected-model manifest is missing.")
SELECTED = read_json(SELECTED_PATH)
_sel_body = {k:v for k,v in SELECTED.items() if k not in {"selected_models_manifest_hash", "created_at_utc"}}
require(sha256_bytes(canonical_json_bytes(_sel_body)) == SELECTED.get("selected_models_manifest_hash"), "Selected-model manifest hash mismatch.")
require(SELECTED.get("upstream_identity") == UPSTREAM_IDENTITY, "Selected models are bound to different upstream artifacts.")
AE_SELECTED = SELECTED["selected_AE"]
DAE_SELECTED = SELECTED["selected_DAE"]

PRIOR_ROOT, PRIOR_STAGE = resolve_dependency_stage('notebook05_threshold_sensitivity')

SENSITIVITY_SEEDS = list(map(int, SEED_SCHEDULE["SENSITIVITY_SEEDS"]))
require(len(SENSITIVITY_SEEDS) == 3, "Exactly three sensitivity seeds are required.")

# Deterministic variants use only already-frozen model-search values/bounds.
def sensitivity_variants(selected, model_type):
    ref = dict(selected["effective_config"])
    variants = []
    # Alternate latent fractions from the frozen search space.
    for v in MODEL_SEARCH_CONTRACT["latent_fraction_candidates"]:
        if float(v) != float(ref["latent_fraction"]):
            q = dict(ref); q["latent_fraction"] = float(v); q["latent_dim"] = latent_dim_for(q["architecture"], q["latent_fraction"])
            variants.append(("latent_fraction", v, q))
    # Frozen search bounds are the prospective LR sensitivity anchors.
    for v in [MODEL_SEARCH_CONTRACT["learning_rate"]["low"], MODEL_SEARCH_CONTRACT["learning_rate"]["high"]]:
        if not math.isclose(float(v), float(ref["learning_rate"]), rel_tol=0, abs_tol=1e-15):
            q = dict(ref); q["learning_rate"] = float(v); variants.append(("learning_rate", v, q))
    for v in MODEL_SEARCH_CONTRACT["weight_decay_candidates"]:
        if float(v) != float(ref["weight_decay"]):
            q = dict(ref); q["weight_decay"] = float(v); variants.append(("weight_decay", v, q))
    if model_type == "DAE":
        for v in MODEL_SEARCH_CONTRACT["dae_corruption_search"][DATASET_NAME]["numerical_masking"]:
            if float(v) != float(ref["numerical_masking"]):
                q = dict(ref); q["numerical_masking"] = float(v); variants.append(("numerical_masking", v, q))
        if DATASET_NAME == "UNSW-NB15":
            for v in MODEL_SEARCH_CONTRACT["dae_corruption_search"][DATASET_NAME]["grouped_categorical_masking"]:
                if float(v) != float(ref["grouped_categorical_masking"]):
                    q = dict(ref); q["grouped_categorical_masking"] = float(v); variants.append(("grouped_categorical_masking", v, q))
    return ref, variants

# Reference results are the frozen Notebook 03 confirmation runs; sensitivity does not reopen selection.
def confirmation_reference(model_type, cfg_hash, seed):
    matches = []
    for p in (CONF_ROOT / "runs").glob("*/run_manifest.json"):
        try:
            m, _ = verify_self_hashed_json(p, "run_manifest_hash")
            if m.get("model_type") == model_type and m.get("config_hash") == cfg_hash and int(m.get("seed")) == int(seed) and m.get("completion_status") == "COMPLETE": matches.append(m)
        except Exception: pass
    require(len(matches)==1, f"Missing/ambiguous confirmation reference {model_type}/{cfg_hash}/seed={seed}")
    return matches[0]

# Prewarm all development-authorized PRE-02 data once. This prevents dozens of
# repeated large Google Drive reads/hashes across AE and DAE sensitivity runs.
prewarm_nonfinal_cache()

# Build the complete consolidated AE + DAE sensitivity execution plan.
_sensitivity_plan = []
_model_plan_summary = []

for _model_type, _selected in [
    ("AE", AE_SELECTED),
    ("DAE", DAE_SELECTED),
]:
    _ref, _variants = sensitivity_variants(
        _selected,
        _model_type,
    )
    _model_plan_summary.append(
        {
            "model_type": _model_type,
            "variant_configurations": len(_variants),
            "sensitivity_seeds": len(SENSITIVITY_SEEDS),
            "new_training_run_slots": len(_variants) * len(SENSITIVITY_SEEDS),
            "reference_runs_reused_from_notebook03": len(SENSITIVITY_SEEDS),
        }
    )
    for _param, _value, _variant in _variants:
        for _seed in SENSITIVITY_SEEDS:
            _sensitivity_plan.append(
                {
                    "model_type": _model_type,
                    "variant_parameter": _param,
                    "variant_value": _value,
                    "seed": int(_seed),
                    "config": _variant,
                }
            )

PLAN_DF = pd.DataFrame(_model_plan_summary)
display(PLAN_DF)

print(
    "UNSW consolidated sensitivity plan — new training run slots:",
    len(_sensitivity_plan),
)
print(
    "Reference runs are reused from Notebook 03 and are not retrained."
)

# Resume audit: completed Notebook 06 run manifests already on Drive are
# authoritative and reused by execute_training_run(..., allow_reuse=True).
_existing_complete = collect_complete_run_manifests(
    STAGE_ROOT,
    stage=NOTEBOOK_STAGE,
)
print(
    "Resume audit — validated COMPLETE sensitivity runs already available:",
    len(_existing_complete),
)
if _existing_complete:
    _resume_df = pd.DataFrame(
        [
            {
                "model_type": m["model_type"],
                "seed": int(m["seed"]),
                "config_hash": m["config_hash"],
                "attempt": int(m.get("attempt", 1)),
                "best_epoch": int(m["best_epoch"]),
                "training_duration_seconds": float(
                    m["training_duration_seconds"]
                ),
            }
            for m in _existing_complete
        ]
    ).sort_values(["model_type", "config_hash", "seed"])
    display(_resume_df.head(100))
del _existing_complete

# Reference results remain frozen Notebook 03 confirmation runs.
rows = []
for model_type, selected in [
    ("AE", AE_SELECTED),
    ("DAE", DAE_SELECTED),
]:
    ref, _ = sensitivity_variants(selected, model_type)
    ref_hash = selected["config_hash"]
    for seed in SENSITIVITY_SEEDS:
        refm = confirmation_reference(
            model_type,
            ref_hash,
            seed,
        )
        rows.append(
            {
                "dataset": DATASET_NAME,
                "model_type": model_type,
                "seed": seed,
                "variant_parameter": "REFERENCE",
                "variant_value": "REFERENCE",
                "config_hash": ref_hash,
                "is_reference": True,
                **refm["metrics"],
            }
        )

# Execute both AE and DAE variants in one resumable consolidated plan.
total_slots = len(_sensitivity_plan)
for slot_index, item in enumerate(_sensitivity_plan, start=1):
    model_type = item["model_type"]
    param = item["variant_parameter"]
    value = item["variant_value"]
    seed = item["seed"]
    variant = item["config"]

    selected = (
        AE_SELECTED
        if model_type == "AE"
        else DAE_SELECTED
    )
    ref = dict(selected["effective_config"])

    # Exactly one intended scientific hyperparameter differs from reference.
    # latent_dim follows latent_fraction deterministically.
    allowed = {param} | (
        {"latent_dim"}
        if param == "latent_fraction"
        else set()
    )
    diffs = {
        k
        for k in set(ref) | set(variant)
        if ref.get(k) != variant.get(k)
    }
    require(
        diffs.issubset(allowed) and param in diffs,
        "Sensitivity variant changed unintended fields: "
        f"target={param}, diffs={sorted(diffs)}",
    )

    print(
        f"[{slot_index}/{total_slots}] "
        f"{model_type} | {param}={value} | seed={seed}"
    )

    m = execute_training_run(
        STAGE_ROOT,
        NOTEBOOK_STAGE,
        model_type,
        variant,
        seed,
        "development",
        allow_reuse=True,
    )

    rows.append(
        {
            "dataset": DATASET_NAME,
            "model_type": model_type,
            "seed": seed,
            "variant_parameter": param,
            "variant_value": value,
            "config_hash": m["config_hash"],
            "is_reference": False,
            **m["metrics"],
        }
    )

results = pd.DataFrame(rows)

# Original stage-compatible outputs.
atomic_write_csv(
    STAGE_ROOT / "sensitivity_seed_results.csv",
    results,
)

summary = (
    results.groupby(
        [
            "model_type",
            "variant_parameter",
            "variant_value",
            "is_reference",
        ],
        dropna=False,
    )
    .agg(
        n_seeds=("seed", "nunique"),
        mean_MCC=("MCC", "mean"),
        median_MCC=("MCC", "median"),
        sd_MCC=("MCC", "std"),
        min_MCC=("MCC", "min"),
        max_MCC=("MCC", "max"),
        mean_FPR=("FPR", "mean"),
        mean_FP_per_1000=("FP_per_1000_benign", "mean"),
        mean_Recall=("Recall", "mean"),
        mean_F1=("F1", "mean"),
        mean_Precision=("Precision", "mean"),
        mean_PR_AUC=("PR-AUC", "mean"),
        mean_Balanced_Accuracy=("Balanced_Accuracy", "mean"),
    )
    .reset_index()
)

# Add reference-relative descriptive deltas without reopening model selection.
reference_means = (
    summary.loc[summary["is_reference"]]
    .set_index("model_type")[
        [
            "mean_MCC",
            "mean_FPR",
            "mean_FP_per_1000",
            "mean_PR_AUC",
        ]
    ]
    .rename(
        columns={
            "mean_MCC": "reference_mean_MCC",
            "mean_FPR": "reference_mean_FPR",
            "mean_FP_per_1000": "reference_mean_FP_per_1000",
            "mean_PR_AUC": "reference_mean_PR_AUC",
        }
    )
)

summary = summary.merge(
    reference_means,
    left_on="model_type",
    right_index=True,
    how="left",
)

summary["delta_mean_MCC_vs_reference"] = (
    summary["mean_MCC"]
    - summary["reference_mean_MCC"]
)
summary["delta_mean_FPR_vs_reference"] = (
    summary["mean_FPR"]
    - summary["reference_mean_FPR"]
)
summary["delta_mean_FP_per_1000_vs_reference"] = (
    summary["mean_FP_per_1000"]
    - summary["reference_mean_FP_per_1000"]
)
summary["delta_mean_PR_AUC_vs_reference"] = (
    summary["mean_PR_AUC"]
    - summary["reference_mean_PR_AUC"]
)

atomic_write_csv(
    STAGE_ROOT / "sensitivity_summary.csv",
    summary,
)

# Explicit one-file-pair consolidated UNSW outputs requested by the researcher.
CONSOLIDATED_SEED_PATH = (
    STAGE_ROOT
    / "UNSW_NB15_AE_DAE_sensitivity_consolidated_seed_results.csv"
)
CONSOLIDATED_SUMMARY_PATH = (
    STAGE_ROOT
    / "UNSW_NB15_AE_DAE_sensitivity_consolidated_summary.csv"
)
atomic_write_csv(CONSOLIDATED_SEED_PATH, results)
atomic_write_csv(CONSOLIDATED_SUMMARY_PATH, summary)

consolidated_substantive = {
    "dataset": DATASET_NAME,
    "stage": NOTEBOOK_STAGE,
    "analysis_scope": "AE + DAE consolidated post-selection OFAT sensitivity",
    "sensitivity_seeds": SENSITIVITY_SEEDS,
    "planned_new_training_run_slots": int(total_slots),
    "reference_runs_reused_from_notebook03": int(
        2 * len(SENSITIVITY_SEEDS)
    ),
    "one_factor_at_a_time_asserted": True,
    "sensitivity_used_for_reselection": False,
    "selected_AE_config_hash_remains": AE_SELECTED["config_hash"],
    "selected_DAE_config_hash_remains": DAE_SELECTED["config_hash"],
    "artifact_hashes": {
        CONSOLIDATED_SEED_PATH.name: sha256_file(
            CONSOLIDATED_SEED_PATH
        ),
        CONSOLIDATED_SUMMARY_PATH.name: sha256_file(
            CONSOLIDATED_SUMMARY_PATH
        ),
    },
}
consolidated_manifest = {
    **consolidated_substantive,
    "created_at_utc": datetime.now(timezone.utc).isoformat(),
    "consolidated_manifest_hash": sha256_bytes(
        canonical_json_bytes(consolidated_substantive)
    ),
}
atomic_write_json(
    STAGE_ROOT / "UNSW_NB15_sensitivity_consolidated_manifest.json",
    consolidated_manifest,
)

stage_substantive = {
    "dataset": DATASET_NAME,
    "stage": NOTEBOOK_STAGE,
    "notebook_version": NOTEBOOK_VERSION,
    "code_version": CODE_VERSION,
    "upstream_identity": UPSTREAM_IDENTITY,
    "sensitivity_seeds": SENSITIVITY_SEEDS,
    "one_factor_at_a_time_asserted": True,
    "sensitivity_used_for_reselection": False,
    "selected_AE_config_hash_remains": AE_SELECTED["config_hash"],
    "selected_DAE_config_hash_remains": DAE_SELECTED["config_hash"],
    "consolidated_AE_DAE_analysis": True,
    "key_artifact_hashes": {
        "sensitivity_seed_results.csv": sha256_file(
            STAGE_ROOT / "sensitivity_seed_results.csv"
        ),
        "sensitivity_summary.csv": sha256_file(
            STAGE_ROOT / "sensitivity_summary.csv"
        ),
        CONSOLIDATED_SEED_PATH.name: sha256_file(
            CONSOLIDATED_SEED_PATH
        ),
        CONSOLIDATED_SUMMARY_PATH.name: sha256_file(
            CONSOLIDATED_SUMMARY_PATH
        ),
        "UNSW_NB15_sensitivity_consolidated_manifest.json": sha256_file(
            STAGE_ROOT
            / "UNSW_NB15_sensitivity_consolidated_manifest.json"
        ),
    },
    "final_status": "PASS",
    "next_authorized_stage": (
        "Notebook 07 final known-attack evaluation"
    ),
}

write_stage_manifest(
    STAGE_ROOT / "stage_manifest.json",
    stage_substantive,
)
create_artifact_inventory(
    STAGE_ROOT,
    STAGE_ROOT / "artifact_inventory.csv",
)

display(summary)
print(
    "PASS: consolidated UNSW AE+DAE hyperparameter sensitivity completed "
    "as robustness analysis only; selected configurations remain frozen."
)
```
